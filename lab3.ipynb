```python
import pandas as pd

data = [
    ["Sunny", "Warm", "Normal", "Strong", "Yes"],
    ["Sunny", "Warm", "High", "Strong", "Yes"],
    ["Rainy", "Cold", "High", "Strong", "No"],
    ["Sunny", "Warm", "High", "Weak", "Yes"]
]

columns = ["Sky", "Temperature", "Humidity", "Wind", "PlayTennis"]

df = pd.DataFrame(data, columns=columns)

print("Dataset:")
print(df)

h = ["Φ", "Φ", "Φ", "Φ"]

print("\nInitial Hypothesis:", h)

for i in range(len(df)):
    if df.iloc[i]["PlayTennis"] == "Yes":
      for j in range(len(h)):
        if h[j] == "Φ":
          h[j] = df.iloc[i, j]
        elif h[j] != df.iloc[i, j]:
          h[j] = "?"

      print("After example", i + 1, ":", h)
print("Final Hypothesis:", h)
```

```python
import pandas as pd

# Load Car Evaluation dataset directly
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/car/car.data"
columns = ['Buying', 'Maint', 'Doors', 'Persons', 'LugBoot', 'Safety', 'Target']
df = pd.read_csv(url, names=columns)

# Convert multi-class target to binary Target ("Yes"/"No")
df['Target'] = df['Target'].apply(lambda x: 'Yes' if x in ['acc', 'good', 'vgood'] else 'No')

# Initialize hypothesis vector with null symbol (Φ)
num_features = len(columns) - 1
h = ['Φ'] * num_features

print("Initial Hypothesis:", h)

# Find-S Algorithm matching your exact loop style
for i in range(len(df)):
    if df.iloc[i]["Target"] == "Yes":
        for j in range(len(h)):
            if h[j] == "Φ":
                h[j] = df.iloc[i, j]
            elif h[j] != df.iloc[i, j]:
                h[j] = "?"

        print("After example", i + 1, ":", h)

print("\nFinal Hypothesis:", h)
```
